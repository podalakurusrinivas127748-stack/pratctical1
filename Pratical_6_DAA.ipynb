```python
import time

def print_optimal_parenthesis(split, i, j):
    if i == j:
        return f"A{i}"

    k = split[i][j]

    left = print_optimal_parenthesis(split, i, k)
    right = print_optimal_parenthesis(split, k + 1, j)

    return f"({left}{right})"


# Number of matrices
n = int(input("Enter number of matrices: "))

# Matrix dimensions
print(f"Enter {n + 1} dimensions:")
print("Example: For A1 = 10x20, A2 = 20x30, A3 = 30x40")
print("Enter: 10 20 30 40")

p = list(map(int, input().split()))

# Check input
if len(p) != n + 1:
    print("Error: You must enter exactly", n + 1, "dimensions.")
    exit()

# DP table
dp = [[0] * (n + 1) for _ in range(n + 1)]

# Split table
split = [[0] * (n + 1) for _ in range(n + 1)]

# Start execution time
start_time = time.perf_counter()

# Matrix Chain Multiplication
for length in range(2, n + 1):

    for i in range(1, n - length + 2):

        j = i + length - 1

        dp[i][j] = float('inf')

        for k in range(i, j):

            cost = (
                dp[i][k]
                + dp[k + 1][j]
                + p[i - 1] * p[k] * p[j]
            )

            if cost < dp[i][j]:
                dp[i][j] = cost
                split[i][j] = k

# End execution time
end_time = time.perf_counter()

execution_time = end_time - start_time

# Results
print("\n--------------------------------------")
print("Minimum number of scalar multiplications:",
      dp[1][n])

print("Optimal Parenthesization:",
      print_optimal_parenthesis(split, 1, n))

print("Execution Time:",
      execution_time, "seconds")

print("Execution Time:",
      execution_time * 1000, "milliseconds")
print("--------------------------------------")

# Display DP table
print("\nDP Table:")

for i in range(1, n + 1):
    for j in range(1, n + 1):

        if j < i:
            print("-", end="\t")
        else:
            print(dp[i][j], end="\t")

    print()
```
