```python
# Kruskal's Algorithm

V = 4

edges = [
    (10, 0, 1),
    (6, 0, 2),
    (5, 0, 3),
    (15, 1, 3),
    (4, 2, 3)
]

# Sort edges by weight
edges.sort()

parent = list(range(V))

def find(i):
    while parent[i] != i:
        i = parent[i]
    return i

def union(i, j):
    parent[find(i)] = find(j)

mst = []
cost = 0

for weight, u, v in edges:
    if find(u) != find(v):
        mst.append((u, v, weight))
        cost += weight
        union(u, v)

print("Minimum Spanning Tree:")

for u, v, weight in mst:
    print(u, "-", v, "=", weight)

print("Minimum Cost:", cost)
```
